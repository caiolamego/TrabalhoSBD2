## Definição do Schema Gold (DDL)
Criação do schema `gold` e as tabelas dimensões e fato:
- **DIM_PAI**: Dimensão País
- **DIM_TMP**: Dimensão Tempo
- **DIM_IND**: Dimensão Indicador
- **FAT_OBS_ECO**: Fato Observações Econômicas

```python
# Criar schema gold e tabelas do Data Warehouse (Gold)
print("Criando schema gold e tabelas do Data Warehouse...")

# SQLs de criação
sql_create_schema = "CREATE SCHEMA IF NOT EXISTS gold;"

sql_create_dim_pais = """
CREATE TABLE IF NOT EXISTS gold.DIM_PAI (
    srk_pai SERIAL PRIMARY KEY,
    cod_pai VARCHAR(10) NOT NULL UNIQUE,
    nom_pai VARCHAR(100)
);
"""

sql_create_dim_tempo = """
CREATE TABLE IF NOT EXISTS gold.DIM_TMP (
    srk_tmp SERIAL PRIMARY KEY,
    cod_per VARCHAR(20) NOT NULL UNIQUE,
    num_ano INTEGER,
    cod_tri VARCHAR(5)
);
"""

sql_create_dim_indicador = """
CREATE TABLE IF NOT EXISTS gold.DIM_IND (
    srk_ind SERIAL PRIMARY KEY,
    cod_ind VARCHAR(100) NOT NULL UNIQUE,
    nom_ind VARCHAR(255),
    nom_fnt VARCHAR(50),
    des_cat VARCHAR(100)
);
"""

sql_create_fato = """
CREATE TABLE IF NOT EXISTS gold.FAT_OBS_ECO (
    srk_pai INTEGER NOT NULL,
    srk_tmp INTEGER NOT NULL,
    srk_ind INTEGER NOT NULL,
    vlr_obs DECIMAL(30, 8),
    PRIMARY KEY (srk_pai, srk_tmp, srk_ind),
    FOREIGN KEY (srk_pai) REFERENCES gold.DIM_PAI(srk_pai),
    FOREIGN KEY (srk_tmp) REFERENCES gold.DIM_TMP(srk_tmp),
    FOREIGN KEY (srk_ind) REFERENCES gold.DIM_IND(srk_ind)
);
"""

sql_truncate_dims = """
TRUNCATE TABLE gold.DIM_PAI, gold.DIM_TMP, gold.DIM_IND RESTART IDENTITY CASCADE;
"""

conn = None
cursor = None
try:
    conn = psycopg2.connect(
        host=DB_HOST,
        port=DB_PORT,
        dbname=DB_NAME,
        user=DB_USER,
        password=DB_PASS
    )
    conn.autocommit = True
    cursor = conn.cursor()
    
    # Criar schema
    print("Criando schema gold...")
    cursor.execute(sql_create_schema)
    
    # Criar tabelas dimensões
    print("Criando gold.Dim_Pais...")
    cursor.execute(sql_create_dim_pais)
    
    print("Criando gold.Dim_Tempo...")
    cursor.execute(sql_create_dim_tempo)
    
    print("Criando gold.Dim_Indicador...")
    cursor.execute(sql_create_dim_indicador)
    
    # Criar tabela fato
    print("Criando gold.Fato_ObservacaoEconomica...")
    cursor.execute(sql_create_fato)
    
    print("Tabelas criadas com sucesso!")
    
    # Limpar tabelas se já existirem (verifica antes de truncar)
    print("\nVerificando e limpando tabelas do gold...")
    
    # Verificar se as tabelas existem antes de truncar
    cursor.execute("""
        SELECT EXISTS (
            SELECT FROM information_schema.tables 
            WHERE table_schema = 'gold' 
            AND table_name = 'DIM_PAI'
        );
    """)
    tables_exist = cursor.fetchone()[0]
    
    if tables_exist:
        print("Executando TRUNCATE nas dimensões...")
        cursor.execute(sql_truncate_dims)
        print("Tabelas do gold truncadas com sucesso.")
    else:
        print("Tabelas ainda não existiam, nenhum TRUNCATE necessário.")
    
except Exception as e:
    print(f"ERRO ao criar/truncar tabelas: {e}")
    raise e
finally:
    if cursor:
        cursor.close()
    if conn:
        conn.close()
```

```python
# Ler dados da tabela silver.bop_clean
print("Lendo dados da tabela silver.bop_clean...")
df_silver = spark.read.jdbc(
    url=DB_URL_JDBC, 
    table="silver.bop_clean", 
    properties=DB_PROPERTIES
)

print(f"Total de registros: {df_silver.count()}")
print(f"Total de colunas lidas: {len(df_silver.columns)}")

# Renomear colunas: substituir "_" por "/" em posições específicas
# Regras:
# 1. Colunas que começam com IRFCL não devem ser alteradas
# 2. Para as demais colunas, trocar "_" por "/" antes e depois de: BOP, IIP, DM, ER

import re

for old_col in df_silver.columns:
    # Pular colunas que começam com IRFCL (case insensitive)
    if old_col.upper().startswith('IRFCL'):
        continue
    
    # Aplicar a transformação: trocar "_" por "/" antes e depois de BOP, IIP, DM, ER
    new_col = old_col
    
    # Padrões: _BOP_, _IIP_, _DM_, _ER_ ou _BOP$, _IIP$, _DM$, _ER$ ou ^BOP_, ^IIP_, ^DM_, ^ER_
    # Trocar _ por / ao redor dessas palavras-chave
    new_col = re.sub(r'_(BOP|IIP|DM|ER)_', r'/\1/', new_col, flags=re.IGNORECASE)
    new_col = re.sub(r'_(BOP|IIP|DM|ER)$', r'/\1', new_col, flags=re.IGNORECASE)
    new_col = re.sub(r'^(BOP|IIP|DM|ER)_', r'\1/', new_col, flags=re.IGNORECASE)
    
    if new_col != old_col:
        df_silver = df_silver.withColumnRenamed(old_col, new_col)

all_columns = df_silver.columns
print(f"Colunas renomeadas. Exemplo das primeiras 5: {all_columns}")
```

### 2. ETL da Dimensão País (DIM_PAI)
Processamento dos países únicos, enriquecimento com nomes em português e carga na tabela `DIM_PAI`.

```python
# ETL Dim_Pais
print("Iniciando ETL para gold.DIM_PAI...")

df_paises_silver = df_silver.select(col("COUNTRY").alias("cod_pai")).distinct() 

schema_pais_map = StructType([
    StructField("cod_pai_map", StringType(), False),
    StructField("nom_pai", StringType(), True)
])

pais_data = [(cod, nome) for cod, nome in pais_map_data]
df_pais_map = spark.createDataFrame(pais_data, schema=schema_pais_map)

df_dim_pais = df_paises_silver.join(
    df_pais_map, 
    df_paises_silver.cod_pai == df_pais_map.cod_pai_map, 
    "left"
).select("cod_pai", "nom_pai")

print(f"Escrevendo {df_dim_pais.count()} países em gold.DIM_PAI...")
df_dim_pais.write.jdbc(
    url=DB_URL_JDBC,
    table="gold.DIM_PAI",
    mode="append",
    properties=DB_PROPERTIES
)
print("gold.DIM_PAI populada com sucesso.")
```

### 3. ETL da Dimensão Tempo (DIM_TMP)
Extração de períodos únicos e quebra em Ano (`num_ano`) e Trimestre (`cod_tri`) para carga na tabela `DIM_TMP`.

```python
# ETL Dim_Tempo
print("Iniciando ETL para gold.DIM_TMP...")

df_dim_tempo = (
    df_silver.select(col("TIME_PERIOD").alias("cod_per")) 
    .distinct()
    .filter(col("cod_per").isNotNull())
)

split_col = split(col("cod_per"), "-")

df_dim_tempo = df_dim_tempo.withColumn("num_ano", split_col.getItem(0).cast("int"))
df_dim_tempo = df_dim_tempo.withColumn(
    "cod_tri",
    when(split_col.getItem(1).isNotNull(), split_col.getItem(1)).otherwise(lit(None))
)

df_dim_tempo_final = df_dim_tempo.select("cod_per", "num_ano", "cod_tri")

print(f"Escrevendo {df_dim_tempo_final.count()} períodos em gold.DIM_TMP...") 
df_dim_tempo_final.write.jdbc(
    url=DB_URL_JDBC,
    table="gold.DIM_TMP",
    mode="append",
    properties=DB_PROPERTIES
)
print("gold.DIM_TMP populada com sucesso.")
```

### 4. ETL da Dimensão Indicador (DIM_IND)
Catalogação de todos os códigos de indicadores, mapeamento de nomes amigáveis e carga na tabela `DIM_IND`.

```python
# ETL Dim_Indicador
print("Iniciando ETL para gold.DIM_IND...")

id_vars = [
    'COUNTRY', 'TIME_PERIOD', 'FREQUENCY', 'UNIT_BOP', 
    'TYPE_OF_TRANSFORMATION_ER', 'UNIT_IIP', 'SECTOR_IRFCL', 
    'TERRITORIAL_LEVEL', 'FREQ'
]

# transformar em minúsculas
id_vars_lower = {x.lower() for x in id_vars}

# 2. Criar dicionário de mapeamento garantindo que a CHAVE seja UPPERCASE
# Isso garante que mesmo se a coluna vier minúscula, vamos achar a chave certa
indicator_map_dict = {cod.upper(): (nome, fonte, cat) for cod, nome, fonte, cat in indicator_map_data}


indicator_codes = [c for c in all_columns if c.lower() not in id_vars_lower]

schema_indicador_data = []

for cod in indicator_codes:
    nome, fonte, cat = indicator_map_dict.get(
        cod.upper(), 
        (f"Indicador - {cod}", "Desconhecida", "Desconhecida")
    )
    schema_indicador_data.append((cod, nome, fonte, cat))

# Definição do Schema com Mnemônicos
schema_indicador = StructType([
    StructField("cod_ind", StringType(), False),
    StructField("nom_ind", StringType(), True),
    StructField("nom_fnt", StringType(), True),
    StructField("des_cat", StringType(), True),
])

df_dim_indicador = spark.createDataFrame(schema_indicador_data, schema=schema_indicador)

print(f"Escrevendo {df_dim_indicador.count()} indicadores em gold.DIM_IND...")
df_dim_indicador.write.jdbc(
    url=DB_URL_JDBC,
    table="gold.DIM_IND",
    mode="append",
    properties=DB_PROPERTIES
)
print("gold.DIM_IND populada com sucesso.")
```

### 5. ETL da Fato Observação Econômica (FAT_OBS_ECO)
Esta etapa realiza o *Unpivot* (transformação de colunas em linhas) dos dados métricos.
Posteriormente, realiza os *Lookups* (joins) com as dimensões já carregadas para recuperar as Surrogate Keys (`srk`) e carregar a tabela fato.

```python
# ETL Fato_ObservacaoEconomica
print("Iniciando ETL para gold.FAT_OBS_ECO...")

# ---------------------------------------------------------
# 1. LIMPEZA PREVENTIVA (TRUNCATE)
# Garante que a tabela esteja vazia antes de carregar, evitando duplicação
# ---------------------------------------------------------
print("Limpando dados antigos da FAT_OBS_ECO...")
conn = psycopg2.connect(host=DB_HOST, port=DB_PORT, dbname=DB_NAME, user=DB_USER, password=DB_PASS)
conn.autocommit = True
cursor = conn.cursor()
try:
    cursor.execute("TRUNCATE TABLE gold.FAT_OBS_ECO RESTART IDENTITY;")
    print("Tabela gold.FAT_OBS_ECO truncada com sucesso.")
except Exception as e:
    print(f"Aviso ao truncar: {e}")
finally:
    cursor.close()
    conn.close()

# ---------------------------------------------------------
# 2. LEITURA E PROCESSAMENTO
# ---------------------------------------------------------
print("Lendo Dimensões do gold (com SRKs)...")

# Leitura das tabelas com nomes mnemônicos oficiais
df_dim_pais = spark.read.jdbc(url=DB_URL_JDBC, table="gold.DIM_PAI", properties=DB_PROPERTIES)
df_dim_tempo = spark.read.jdbc(url=DB_URL_JDBC, table="gold.DIM_TMP", properties=DB_PROPERTIES)
df_dim_indicador = spark.read.jdbc(url=DB_URL_JDBC, table="gold.DIM_IND", properties=DB_PROPERTIES)

# Colunas de identificação
id_vars = ['country', 'time_period']

# Identificar colunas metadados (strings) vs valores (numéricos)
from pyspark.sql.types import StringType
measure_vars = [field.name for field in df_silver.schema.fields 
                if not isinstance(field.dataType, StringType) and field.name not in id_vars]

# Stack expression
stack_expr_parts = []
for cod in measure_vars:
    stack_expr_parts.append(f"'{cod}', `{cod}`") 
        
stack_expr = f"stack({len(measure_vars)}, {', '.join(stack_expr_parts)}) as (cod_ind_origem, Valor)"

df_long = df_silver.selectExpr(*id_vars, stack_expr)
df_long_cleaned = df_long.filter(col("Valor").isNotNull())

print("Juntando Fato com Dimensões para obter SRKs...")

# Renomeando colunas da Silver para preparar para o Join
df_long_renomeado = df_long_cleaned.withColumnRenamed("country", "ref_cod_pai") \
                                 .withColumnRenamed("time_period", "ref_cod_per") \
                                 .withColumnRenamed("cod_ind_origem", "ref_cod_ind")

# Joins
df_fato_join_pais = df_long_renomeado.join(df_dim_pais, df_long_renomeado.ref_cod_pai == df_dim_pais.cod_pai, "inner") \
    .select(df_long_renomeado["*"], df_dim_pais.srk_pai)

df_fato_join_tempo = df_fato_join_pais.join(df_dim_tempo, df_fato_join_pais.ref_cod_per == df_dim_tempo.cod_per, "inner") \
    .select(df_fato_join_pais["*"], df_dim_tempo.srk_tmp)

df_fato_join_final = df_fato_join_tempo.join(df_dim_indicador, df_fato_join_tempo.ref_cod_ind == df_dim_indicador.cod_ind, "inner") \
    .select(df_fato_join_tempo["*"], df_dim_indicador.srk_ind)

# Seleção Final
df_fato_final = df_fato_join_final.select(
    col("srk_pai"),
    col("srk_tmp"),
    col("srk_ind"),
    col("Valor").cast("decimal(30, 8)").alias("vlr_obs")
)

# ---------------------------------------------------------
# 3. DEDUPLICAÇÃO E ESCRITA
# remove duplicatas exatas se houver sujeira na origem
# ---------------------------------------------------------
df_fato_final = df_fato_final.dropDuplicates(["srk_pai", "srk_tmp", "srk_ind"])

print(f"Escrevendo {df_fato_final.count()} linhas únicas em gold.FAT_OBS_ECO...")

df_fato_final.write.jdbc(
    url=DB_URL_JDBC,
    table="gold.FAT_OBS_ECO",
    mode="append", # Mantemos append, pois já fizemos o TRUNCATE manual acima
    properties=DB_PROPERTIES
)
print("gold.FAT_OBS_ECO populada com sucesso.")
```
